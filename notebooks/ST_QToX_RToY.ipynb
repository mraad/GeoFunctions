# ST_QToX(q,cell,dist=0.0), ST_RToY(r,cell,dist=0.0)

Convert q or r value to x or y mercator meter values.

```python
import geofunctions as S
import pyspark.sql.functions as F
```

```python
cell = 10.0
```

```python
(spark
 .range(5)
 .withColumn("lon", F.rand() * 360 - 180)
 .withColumn("q", S.st_lontoq("lon", cell))
 .withColumn("x", S.st_qtox("q", cell))
 .show(truncate=False)
)
```

```python
(spark
 .range(5)
 .withColumn("lat", F.rand() * 180 - 90)
 .withColumn("r", S.st_lattor("lat", cell))
 .withColumn("y", S.st_rtoy("r", cell))
 .show(truncate=False)
)
```
