````python
from osgeo import ogr
from osgeo.gdalconst import GA_ReadOnly
````

````python
ogr.UseExceptions()
````

````python
driver = ogr.GetDriverByName("OpenFileGDB")
if driver is None:
    print("No support for OpenFileGDB")
````

````python
gdb_path = "../data/Miami.gdb"
ds = driver.Open(gdb_path, GA_ReadOnly)  # 0 means read-only, 1 means writeable
````

````python
rows = ds.ExecuteSQL("select count(*) as count from Broadcast where Status = 0")
for row in rows:
    print(row.GetFID(), row.GetFieldAsInteger(0))
````

````python
layer = ds.GetLayerByName("Broadcast")
````

````python
feature_count = layer.GetFeatureCount()
feature_count
````

````python
seeks = list(range(0, feature_count, feature_count // 10))
````

````python
seeks[-1] += feature_count - seeks[-1]
````

````python
counts = list(zip(seeks[0:], seeks[1:]))
````

````python
counts = list(map(lambda x: x[1] - x[0], counts))
````

````python
seeks_counts = list(zip(seeks, counts))
# seeks_counts
````

````python
df = spark.createDataFrame(seeks_counts, ["seek", "count"])
````
