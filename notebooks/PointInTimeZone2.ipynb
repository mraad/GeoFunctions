```python
cell = 15.0  # QR in degrees.
dist = 0.00001  # Padding in degrees.
wkid = 4326  # SR ID
```

### Generate random points.

```python
xmin, ymin, xmax, ymax = (-180.0, -80.0, 180.0, 80.0)
xdel = xmax - xmin
ydel = ymax - ymin

lhs = (
    spark.range(0, 100_000)
    .repartition(200)
    .withColumn("x", F.rand() * xdel + xmin)
    .withColumn("y", F.rand() * ydel + ymin)
    .select(
        S.st_point("x", "y"),
    )
    .withColumn("qr", S.qr_envp_explode("geom", cell))
    .repartition(200, "qr.qr")
    .cache()
)
```

```python
lhs.printSchema()
```

### Read the time zone polygons.

```python
rhs = (
    spark.read.format("shp")
    .options(path="../data/world/tz_world.shp")
    .load()
    .repartition(200)
    .withColumnRenamed("shape", "geom")
    .withColumn("qr", S.qr_envp_explode("geom", cell, dist))
    .repartition(200, "qr.qr")
    .cache()
)
```

```python
rhs.printSchema()
```

### Perform Point in Polygon.

```python
df = (
    lhs.withColumnRenamed("geom", "geometry")
    .join_qr(
        rhs,
        cell,
        oper="within",
        wkid=wkid,
        lhs_geom="geometry",
    )
    .drop("geom")
    # .printSchema()
)
```

### Convert to GeoPandas and visualize.

```python
%%time

pdf = df.toPandas()
pdf.geometry = gp.GeoSeries.from_wkb(pdf.geometry)
gdf = gp.GeoDataFrame(pdf, crs=f"EPSG:{wkid}")
_ = gdf.sindex
# Enable the spatial index.
```

```python
gdf.explore(tiles=xyz.Esri.WorldGrayCanvas)
```
