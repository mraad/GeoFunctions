# QR_Envp(geom,cell,dist=0.0)

Convert a geometry to its qr / envp elements. This is typically used with `QR_INTERSECT` (lower/left intersection check)

```python
import geofunctions as S
import geopandas as gp
import matplotlib.pyplot as plt
import pyspark.sql.functions as F
```

```python
data = [("POLYGON((0 0,100 0,100 100,0 0))",)]
```

```python
df = spark.createDataFrame(data, "wkt string").select(
    S.st_fromtext("wkt").alias("geometry")
)
```

```python
cell = 40.0
dist = 0.0

qr = df.select(F.explode(S.qr_envp(df.geometry, cell, dist)).alias("qr"))
```

### Note that xmin,ymin,xmax,ymax represent the envelope of the geometry in each QR.

```python
qr.printSchema()
qr.count()
```

```python
qr.select(
    F.hex("qr.qr").alias("qr_hex"),
    "qr.xmin",
    "qr.ymin",
    "qr.xmax",
    "qr.ymax",
).toPandas()
```
