# QR_FromXY(x,y,cell)

Convert X,Y to QR value within a QR cell.

```python
import geofunctions as S
import pyspark.sql.functions as F
```

```python
data = [(123.4, 56.8)]
```

```python
(spark
     .createDataFrame(data, "x double, y double")
     .select(F.hex(S.qr_fromxy("x","y", cell=50.0)).alias("qr_hex"))
     .show(truncate=False)
)
```
