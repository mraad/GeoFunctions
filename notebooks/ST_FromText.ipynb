# ST_FromText(wkt)

Create geomerty from WKT string.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
data = [("POLYGON ((30 10, 40 40, 20 40, 10 20, 30 10))",)]
```

```python
pdf = spark.createDataFrame(data,"wkt string").select(S.st_fromtext("wkt").alias("geometry")).toPandas()
```

```python
pdf.geometry = gp.GeoSeries.from_wkb(pdf.geometry)
gdf = gp.GeoDataFrame(pdf)
_ = gdf.plot(linewidth=1, edgecolor="black", color="red")
```
