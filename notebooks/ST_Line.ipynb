# ST_Line(x1,y1,x2,y2)

Create a line geometry from given coordinates.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
(
    spark.range(3)
    .withColumn("x1", F.rand())
    .withColumn("y1", F.rand())
    .withColumn("x2", F.rand())
    .withColumn("y2", F.rand())
    .select("id", S.st_astext(S.st_line("x1", "y1", "x2", "y2")).alias("line"))
    .show(truncate=False, vertical=True)
)
```
