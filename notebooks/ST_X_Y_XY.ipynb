# ST_X(geom,index=0), ST_Y(geom,index=0), ST_XY(geom,index=0)

Return the x or y or (x,y) of a geometry at an index position.

```python
import geofunctions as S
import pyspark.sql.functions as F
```

```python
data = [("LINESTRING(0 0,50 200, 100 50)",)]
```

```python
df = spark.createDataFrame(data, "wkt string").select(S.st_fromtext("wkt").alias("geom"))
```

```python
xy = df.select(
    S.st_x(df.geom,index=0).alias("x0"),
    S.st_y(df.geom, 1).alias("y1"),
    S.st_xy(df.geom, -1).alias("last xy")
)

xy.printSchema()
xy.show(truncate=False, vertical=True)
```
