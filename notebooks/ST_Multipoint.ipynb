# ST_Multipoint(\[ST_Point\])

Create a multipoint geometry from an array of point geometry.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
data = [(0.0, 0.0, 50.0, 50.0, 100.0, 20.0)]
```

```python
df = spark.createDataFrame(
    data, "x1 double,y1 double," "x2 double,y2 double," "x3 double,y3 double"
)
```

```python
pdf = df.select(
    S.st_multipoint(
        S.st_point("x1", "y1"),
        S.st_point("x2", "y2"),
        S.st_point("x3", "y3"),
    ).alias("geometry")
).toPandas()
```

```python
pdf.geometry = gp.GeoSeries.from_wkb(pdf.geometry)
gdf = gp.GeoDataFrame(pdf)
_ = gdf.plot()
```
