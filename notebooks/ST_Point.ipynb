# ST_Point(x,y)

Create a point geometry.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
data = [(123.4, 56.7)]
```

```python
(
    spark.createDataFrame(data, "x double,y double")
    .select(S.st_astext(S.st_point("x", "y")).alias("Point"))
    .show(truncate=False)
)
```
