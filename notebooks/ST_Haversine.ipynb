# ST_Haversine(lon1,lat1,lon2,lat2)

Calculate the haversine distance in meters between 2 geographic locations.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
(spark
     .range(3)
     .withColumn("lon1", F.rand() * 180.0 - 90.0)
     .withColumn("lat1", F.rand() * 360.0 - 180.0)
     .withColumn("lon2", F.rand() * 180.0 - 90.0)
     .withColumn("lat2", F.rand() * 360.0 - 180.0)
     .withColumn("haversine", S.st_haversine("lon1","lat1","lon2","lat2"))
     .show(truncate=False, vertical=True)
)
```
