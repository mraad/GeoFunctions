# ST_LonToQ(lon,cell), ST_LonToX(lon)

- Convert longitude value to q given a cell value.
- Convert longitude value to web mercator horizontal meters.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
cell = 2.0
```

```python
(
    spark.range(3)
    .withColumn("lon", F.rand() * 360.0 - 180.0)
    .select("id",
            "lon",
            F.format_number(S.st_lontox("lon"),2).alias("x"),
            S.st_lontoq("lon", cell)
           )
    .show(truncate=False, vertical=True)
)
```
