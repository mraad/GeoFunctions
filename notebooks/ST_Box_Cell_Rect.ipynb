# ST_BOX(x,y,h,v=None) / ST_CELL(x,y,w,h=None) / ST_RECT(x1,y1,x2,y2)

Create rectangle polygon using various inputs.

```python
import geofunctions as S
import pyspark.sql.functions as F
```

```python
data = [
    (10.0, 10.0, 100.0, 50.0),
]
(
    spark.createDataFrame(data, "x1 double,y1 double,x2 double,y2 double")
    .select(S.st_astext(S.st_rect("x1", "y1", "x2", "y2")).alias("geom"))
    .show(truncate=False)
)
```

```python
data = [
    (10.0, 10.0, 100.0, 50.0),
]
(
    spark.createDataFrame(data, "x double,y double,w double,h double")
    .select(S.st_astext(S.st_cell("x", "y", "w", "h")).alias("geom"))
    .show(truncate=False)
)
```

```python
data = [
    (0.0, 0.0, 100.0, 50.0),
]
(
    spark.createDataFrame(data, "x double,y double,h double,v double")
    .select(S.st_astext(S.st_box("x", "y", "h", "v")).alias("geom"))
    .show(truncate=False)
)
```
