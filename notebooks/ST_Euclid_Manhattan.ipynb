# ST_Euclid(x1,y1,x2,y2) / ST_Manhattan(x1,y1,x2,y2)

Calculate euclid or manhattan distance for given (x1,y1) (x2,y2)

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
data = [(0.0, 0.0, 100.0, 100.0)]
```

```python
(
    spark.createDataFrame(data, "x1 double,y1 double,x2 double,y2 double")
    .select(
        S.st_euclid("x1", "y1", "x2", "y2").alias("euclid"),
        S.st_manhattan("x1", "y1", "x2", "y2").alias("manhattan"),
    )
    .show(truncate=False)
)
```
