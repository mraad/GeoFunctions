# ST_Contains(lhs_geom, rhs_geom)

Check if rhs geometry is contained in lhs geometry.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
data = [(10.0, 10.0, 90.0, 60.0, 60.0)]
```

```python
(
    spark.createDataFrame(
        data,
        "x double, y double, cell double,"
        "px double, py double"
    )
    .select(
        S.st_cell("x", "y", "cell").alias("polygon"),
        S.st_point("px", "py").alias("point"),
    )
    .select(S.st_contains("polygon", "point").alias("Is point in polygon?"))
    .show(truncate=False)
)
```
