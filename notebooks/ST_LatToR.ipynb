# ST_LatToR(lat,cell) / ST_LatToY(lat)

Convert lat value to QR r value and convert lat value to web mercator y.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
(
    spark.range(3)
    .withColumn("lat", F.rand() * 180.0 - 90.0)
    .withColumn("y", S.st_lattoy("lat"))
    .withColumn("r", S.st_lattor("lat", cell=10.0))
    .show(truncate=False, vertical=True)
)
```
