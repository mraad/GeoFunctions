# ST_Distance(lhs_geom, rhs_geom)

Calculate the distance between 2 geometries.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
data = [(10.0, 10.0)]

lhs = spark.createDataFrame(data,"x double, y double").select(S.st_point("x","y").alias("lhs"), F.lit(1).alias("on"))
```

```python
data = [(100.0, 100.0)]

rhs = spark.createDataFrame(data,"x double, y double").select(S.st_point("x","y").alias("rhs"), F.lit(1).alias("on"))
```

```python
lhs.join(rhs, on="on").select(S.st_distance(lhs.lhs, rhs.rhs).alias("distance")).show(truncate=False)
```
