# ST_AsText(geom)

Convert a geometry to its WKT representation.

```python
import geofunctions as S
import pyspark.sql.functions as F
```

```python
data = [(10.0, 10.0, 20.0, 20.0, 20.0, 10.0)]
```

```python
(
    spark.createDataFrame(
        data, "x1 double,y1 double,x2 double,y2 double,x3 double,y3 double"
    )
    .select(
        S.st_astext(S.st_polyline2(F.array("x1", "y1", "x2", "y2", "x3", "y3"))).alias(
            "wkt"
        )
    )
    .show(truncate=False)
)
```

```python
data = [(0.0, 0.0, 50.0, 50.0)]
```

```python
df = spark.createDataFrame(data, "lx double, ly double, rx double, ry double").select(
    S.st_cell("lx", "ly", F.lit(100.0)).alias("lhs"),
    S.st_cell("rx", "ry", F.lit(100.0)).alias("rhs"),
)
```

```python
(
    df.select(S.st_astext(S.st_intersection("lhs", "rhs")).alias("geom")).show(
        truncate=False
    )
)
```
