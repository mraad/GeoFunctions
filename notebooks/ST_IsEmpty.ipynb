# ST_IsEmpty(geom)

Check if the geometry is empty.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
data = [
    (None,),
    ("POINT (100 100)",),
]
```

```python
(
    spark.createDataFrame(
        data,
        "text STRING",
    )
    .select(
        F.isnull("text").alias("Is Null"),
        # Implicit 'text' as arg, 'geom' as alias.
        S.st_fromtext(),
    )
    .withColumn(
        "is empty",
        # Implicit 'geom' as arg.
        S.st_isempty(),
    )
    .show()
)
```

```python
data = [
    (
        "POLYGON((0 0,0 40,40 40,40 0,0 0))",
        "POLYGON((50 50,50 100,100 100,100 50,50 50))",
    )
]
```

```python
(
    spark.createDataFrame(data, "lhs string,rhs string")
    .select(
        S.st_isempty(
            S.st_intersection(
                S.st_fromtext("lhs"),
                S.st_fromtext("rhs"),
            )
        ).alias("Is Empty")
    )
    .show(truncate=False)
)
```
