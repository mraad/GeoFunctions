# ST_XToQ(x,cell), ST_YToR(r,cell)

Convert x or y mercator meter values to q or r values.

```python
import geofunctions as S
import pyspark.sql.functions as F
```

```python
cell = 10.0
```

```python
(spark
 .range(5)
 .withColumn("x", S.st_lontox(F.rand() * 360 - 180))
 .withColumn("q", S.st_xtoq("x", cell))
 .show(truncate=False)
)
```

```python
(spark
 .range(5)
 .withColumn("y", S.st_lattoy(F.rand() * 180 - 90))
 .withColumn("r", S.st_ytor("y", cell))
 .show(truncate=False)
)
```
