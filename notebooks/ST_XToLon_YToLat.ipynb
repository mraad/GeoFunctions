# ST_XToLon(x), ST_YToLat(y)

Convert x or y web mercator meters to lon or lat value.

```python
import geofunctions as S
import pyspark.sql.functions as F
```

```python
df = spark.range(3).withColumn("lon", F.rand() * 360 - 180).withColumn("lat", F.rand() * 180 - 90)
```

```python
df.show(truncate=False)
```

```python
df.select(
    S.st_xtolon(S.st_lontox("lon")).alias("lon"),
    S.st_ytolat(S.st_lattoy("lat")).alias("lat"),
).show(truncate=False)
```
