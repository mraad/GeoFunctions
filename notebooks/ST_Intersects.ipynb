# ST_Intersects(lhs_geom,rhs_geom)

Check if 2 geometries intersect.

```python
import geofunctions as S
import geopandas as gp
import pyspark.sql.functions as F
```

```python
data = [(
    "POLYGON((0 0,0 100,100 100,100 0,0 0))",
    "POLYGON((50 50,50 150,150 150,150 50,50 50))"
)]
```

```python
(spark
     .createDataFrame(data,"lhs string,rhs string")
     .select(
             S.st_intersects(
                 S.st_fromtext("lhs"),
                 S.st_fromtext("rhs"),
             ).alias("intersects")
     )
     .show(truncate=False)
)
```
